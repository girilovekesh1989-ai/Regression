```python
'''
Question 1: What is Simple Linear Regression?
'''
'''
Simple linear regression is a statistical method that models the linear relationship between one continuous independent variable (\(x\)) and one continuous dependent variable (\(y\)).
Detailed guides and academic modules are available via Penn State STAT 462 and the Wikipedia Simple Linear Regression Entry.

Key Components of the Model

• Dependent Variable (\(y\)): The target or outcome you want to predict.
• Independent Variable (\(x\)): The single predictor or input feature used to make the prediction.
• Regression Equation: Expressed as \(y = \beta_0 + \beta_1x + \epsilon\), where:
	• \(\beta _{0}\) is the y-intercept (the expected value of \(y\) when \(x = 0\)).
	• \(\beta _{1}\) is the slope (how much \(y\) changes for a one-unit increase in \(x\)).
	• \(\epsilon \) is the error term (the difference or residual between the observed value and the true line).

How It Works

• Best-Fit Line: The model finds a straight line that best represents the data points.
• Method of Least Squares: It calculates the line that minimizes the sum of squared errors (the squared differences between actual and predicted values).
• Common Example: Predicting a worker's salary (\(y\)) based on their years of experience (\(x\)).
'''
```

```python
'''
Question 2: What are the key assumptions of Simple Linear Regression?
'''
'''
The core assumptions of the simple linear regression model include:
• Linearity: The relationship between the independent variable (\(X\)) and the expected mean of the dependent variable (\(Y\)) is a straight line.
• Independence of observations: The error terms (residuals) are independent of each other with no autocorrelation (common in sequential or time-series data).
• Homoscedasticity (Constant variance): The variance of the residuals is the same and remains constant across all levels of the independent variable \(X\).
• Normality of residuals: For any fixed value of \(X\), the error terms or residuals are normally distributed (crucial for hypothesis testing, p-values, and confidence intervals).
• Zero conditional mean of errors: The average or expected value of the error term given any value of \(X\) is equal to zero (\(E(\epsilon \vert{} X) = 0\)).
If you are working on a specific dataset or checking model diagnostics, let me know:
• What kind of data you are analyzing (e.g., cross-sectional, time-series)
• Which diagnostic test or plot (like residuals-versus-fitted or Q-Q plot) you need help interpreting
'''
```

```python
'''
Question 3: What is heteroscedasticity, and why is it important to address in regression
models?
'''
'''

What is Heteroscedasticity?

Heteroscedasticity occurs in regression analysis when the variability of the error terms (residuals) is not constant across different values of the independent variables.
• Meaning: In a homoscedastic (ideal) model, the spread of residuals remains uniform. In a heteroscedastic model, the spread changes—often forming a cone, fan, or bow-tie shape when plotted against predicted or independent values (for example, residuals growing larger as household income or property size increases).
• Real-world cause: It frequently appears in cross-sectional data where a wide range of values causes larger errors at higher magnitudes (e.g., predicting savings based on income, where high-income earners exhibit a much wider variance in spending than low-income earners).

Why is it Important to Address?

Failing to address heteroscedasticity breaks a core assumption of the Classical Linear Regression Model. It leads to specific statistical consequences:
• Unbiased, but Inefficient Estimators: Your Ordinary Least Squares (OLS) slope and intercept coefficients remain unbiased (they still point to the right average relationship). However, they are no longer BLUE (Best Linear Unbiased Estimators) because they do not have the lowest possible variance. Other methods can estimate the relationship with greater precision.
• Invalid Hypothesis Testing and p-values: OLS underestimates or overestimates the true standard errors of the coefficients. As a result, your \(t\)-tests and \(F\)-tests become unreliable, leading to incorrect confidence intervals and a high risk of false positives (Type I errors) or false negatives in your \(p\)-values.
• Misleading Predictions: The prediction intervals for new observations will be too narrow in some regions and too wide in others, reducing model reliability.

How to Identify and Fix It

• Identification:
	• Visual: Check residual vs. fitted plots for cone or fan patterns.
	• Statistical Tests: Use formal diagnostics such as the Breusch-Pagan test or the White test.
• Common Remedies:
	• Transform the data: Apply a logarithmic or Box-Cox transformation to the dependent or independent variables to stabilize variance.
	• Use Weighted Least Squares (WLS): Give lower weights to observations with higher variance.
	• Robust Standard Errors: Apply heteroscedasticity-consistent (sandwich) standard errors (such as Huber-White standard errors) so that your \(t\)-tests and \(p\)-values remain valid without changing the coefficient estimates.
'''
```

```python
'''
Question 4: What is Multiple Linear Regression?
'''
'''
Multiple Linear Regression (MLR) is a statistical technique that uses two or more independent (explanatory) variables to predict the outcome of a single continuous dependent (response) variable.
According to the Investopedia Guide to Multiple Linear Regression, it extends ordinary least-squares (OLS) regression by evaluating the collective and individual influence of multiple factors simultaneously.

The MLR Equation

The general mathematical model is expressed as:
\(y_{i}=\beta _{0}+\beta _{1}x_{i1}+\beta _{2}x_{i2}+\dots +\beta _{p}x_{ip}+\epsilon _{i}\)
• \(y_{i}\): The continuous dependent (predicted) variable.
• β₀: The y-intercept (the expected value of y when all predictors are zero).
• \(\beta_1, \beta_2, \dots, \beta_p\): The slope coefficients; each measures the estimated change in y for a one-unit increase in that specific predictor, holding all other predictors constant.
• \(x_{i1}, x_{i2}, \dots, x_{ip}\): The independent (explanatory) variables (continuous or dummy-coded categorical).
• \(\epsilon _{i}\): The model's error term (residuals).

Key Core Concepts

• Controlling for Variables: Unlike simple linear regression (which looks at only one predictor), MLR isolates the unique effect of each variable while "controlling for" or holding fixed the other variables in the model. For example, predicting salary using both years of experience and education level.
• Key Assumptions: Outlined by resources like Testbook's Multiple Regression Overview, a valid MLR model relies on:
	• Linearity: A linear relationship exists between the predictors and the outcome.
	• Independence of Observations: Residuals are independent of one another.
	• Homoscedasticity: Variance of the errors is constant across all levels of the predictors.
	• Normality: The model's residuals follow a normal distribution.
	• No Multicollinearity: Independent variables should not be too highly correlated with each other.
'''
```

```python
'''
Question 5: What is polynomial regression, and how does it differ from linear
regression?
'''
'''
Polynomial regression is a form of regression analysis that models the relationship between an independent variable \(x\) and a dependent variable \(y\) as an \(n\)-th degree curve (such as a parabola or cubic curve).

What is Polynomial Regression?

• Definition: It fits a non-linear, curved relationship to data points by raising the predictor variable to various powers (\(x, x^2, x^3, \dots, x^n\)).
• Formula: Expressed as \(y = b_0 + b_1x + b_2x^2 + b_3x^3 + \dots + b_nx^n + \epsilon\).
• Key Feature: Even though the relationship between \(x\) and \(y\) is curved, the model is linear in its parameters/coefficients (\(b_0, b_1, b_2, \dots\)), meaning it is solved using standard linear regression techniques.
• Common Uses: Modeling complex phenomena like plant growth cycles, temperature variations over a day, or economic trends that curve over time.

How It Differs from Linear Regression


Feature	                                                Linear Regression	                                        Polynomial Regression
Shape of Line	                                 Straight line (\(y = b_0 + b_1x\))	              Curved or bendy line (\(y = b_0 + b_1x + b_2x^2 + \dots\))
Underlying Trend	                    Assumes a constant rate of change between variables	     Captures accelerating, decelerating, or fluctuating patterns
Complexity & Flexibility	                Simple, rigid, and has low model flexibility	   High flexibility, allowing complex shapes depending on the degree \(n\)
Overfitting Risk	                        Low risk of fitting random noise in the data	    Higher risk of overfitting, especially with high-degree polynomials (\(n\))
Best Used When	                          Data points follow a roughly straight-line trend	        Data points show a clear curve or non-linear trend

'''
```

```python
'''
Question 6: Implement a Python program to fit a Simple Linear Regression model to
the following sample data:
● X = [1, 2, 3, 4, 5]
● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
Plot the regression line over the data points.
'''
'''
Here is the complete Python program using the scikit-learn and matplotlib libraries to fit the simple linear regression model and plot the resulting line over your sample data points.

Python Code
'''

import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Define the sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Reshaped to 2D array as required by scikit-learn
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])

# 2. Initialize and fit the Simple Linear Regression model
model = LinearRegression()
model.fit(X, Y)

# 3. Predict values using the trained model to draw the regression line
Y_pred = model.predict(X)

# 4. Display the calculated parameters
print(f"Learned Slope (m): {model.coef_[0]:.4f}")
print(f"Learned Intercept (b): {model.intercept_:.4f}")
print(f"Regression Equation: Y = {model.coef_[0]:.4f} * X + {model.intercept_:.4f}")

# 5. Plot the actual data points and the fitted regression line
plt.scatter(X, Y, color="blue", label="Actual Data Points", zorder=5)
plt.plot(
    X,
    Y_pred,
    color="red",
    linewidth=2,
    label=f"Regression Line (Y = {model.coef_[0]:.2f}X + {model.intercept_:.2f})",
)

# Labeling the chart
plt.title("Simple Linear Regression Model Fit")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

# Display the plot
plt.show()
```

```python
'''
Question 7: Fit a Multiple Linear Regression model on this sample data:
● Area = [1200, 1500, 1800, 2000]
● Rooms = [2, 3, 3, 4]
● Price = [250000, 300000, 320000, 370000]
'''
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 1. Define the sample data
Area = [1200, 1500, 1800, 2000]
Rooms = [2, 3, 3, 4]
Price = [250000, 300000, 320000, 370000]

# Create a DataFrame for independent variables
X = pd.DataFrame({
    'Area': Area,
    'Rooms': Rooms
})
y = pd.Series(Price)

# 2. Fit the Multiple Linear Regression model
X_const = sm.add_constant(X)
model = sm.OLS(y, X_const).fit()
print("--- Regression Results Summary ---")
print(model.summary())

# 3. Check for Multicollinearity using VIF
vif_data = pd.DataFrame()
vif_data["Feature"] = X_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])]

print("\n--- Variance Inflation Factor (VIF) ---")
print(vif_data.to_string(index=False))
```

```python
'''
Question 8: Implement polynomial regression on the following data:
● X = [1, 2, 3, 4, 5]
3
● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
Fit a 2nd-degree polynomial and plot the resulting curve.
'''
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])

# Fit 2nd-degree polynomial
poly_coefs = np.polyfit(X, Y, 2)
p2 = np.poly1d(poly_coefs)

a, b, c = poly_coefs
print(f"a = {a:.4f}, b = {b:.4f}, c = {c:.4f}")

# Plotting
X_smooth = np.linspace(0.8, 5.2, 200)
Y_smooth = p2(X_smooth)

plt.figure(figsize=(8, 5))
plt.scatter(X, Y, color='red', label='Data Points', s=80, zorder=5)
plt.plot(X_smooth, Y_smooth, color='blue', label=f'2nd-Degree Fit: Y = {a:.2f}X² + {b:.2f}X + {c:.2f}')
plt.title('2nd-Degree Polynomial Regression', fontsize=14)
plt.xlabel('X', fontsize=12)
plt.ylabel('Y', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()
plt.savefig('poly_regression.png', dpi=300)
plt.show()

```

```python
'''
Question 9: Create a residuals plot for a regression model trained on this data:
● X = [10, 20, 30, 40, 50]
● Y = [15, 35, 40, 50, 65]
Assess heteroscedasticity by examining the spread of residuals.
(Include your Python code and output in the code box below.)
'''
import numpy as np
import matplotlib.pyplot as plt

# Data
X = np.array([10, 20, 30, 40, 50])
Y = np.array([15, 35, 40, 50, 65])

# Fit simple linear regression: Y = beta0 + beta1 * X
poly = np.polyfit(X, Y, 1)
Y_pred = np.polyval(poly, X)

# Calculate residuals
residuals = Y - Y_pred

# Output coefficients and values
print(f"Slope (beta1): {poly[0]:.2f}")
print(f"Intercept (beta0): {poly[1]:.2f}")
print("Predicted Y:", Y_pred)
print("Residuals:", residuals)

# Create residual plot
plt.figure(figsize=(8, 5))
plt.scatter(X, residuals, color='blue', s=80, zorder=3)
plt.axhline(0, color='red', linestyle='--', linewidth=1.5, label='Residual = 0 Line')
plt.title('Residuals Plot', fontsize=14)
plt.xlabel('X (Independent Variable)', fontsize=12)
plt.ylabel('Residuals (Observed - Predicted)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('residuals_plot.png', dpi=300)
plt.show()
```

```python
'''
Question 10: Imagine you are a data scientist working for a real estate company. You
need to predict house prices using features like area, number of rooms, and location.
However, you detect heteroscedasticity and multicollinearity in your regression
model. Explain the steps you would take to address these issues and ensure a robust
model.
'''
'''
To address multicollinearity and heteroscedasticity in a house price regression model, you can apply a structured sequence of diagnostic and remediation steps to ensure stability and accuracy.

Step 1: Address Multicollinearity

Multicollinearity happens when predictor variables (such as square footage, total rooms, and number of bedrooms) are strongly correlated with each other, making coefficient estimates unstable.
• Detect the Issue:
	• Correlation Matrix: Inspect a heatmap of Pearson/Spearman correlation coefficients to spot high pairwise correlations (e.g., \(r > 0.8\)).
	• Variance Inflation Factor (VIF): Calculate the VIF for each feature. A VIF value exceeding 5 or 10 indicates severe multicollinearity.
• Remediate the Issue:
	• Drop Redundant Features: Remove one of the highly collinear variables (e.g., drop total rooms if square footage or bedroom count already captures that information).
	• Feature Engineering / Combination: Combine correlated features into a single composite metric (e.g., create an average room size feature instead of keeping total area and total rooms separate).
	• Apply Regularization: Use Ridge Regression (\(\text{L2}\) penalty) or Elastic Net, which shrinks the coefficients of correlated variables and stabilizes the model.
	• Dimensionality Reduction: Use PCA (Principal Component Analysis) to transform correlated features into a set of uncorrelated components before modeling.

Step 2: Address Heteroscedasticity

Heteroscedasticity occurs when the variance of the error terms (residuals) is not constant across all levels of the independent variables—for example, higher-priced or larger luxury homes having much wider prediction errors than smaller apartments.
• Detect the Issue:
	• Residual Plots: Plot the residuals against predicted values or individual features; a "funnel" or cone shape indicates non-constant variance.
	• Formal Tests: Run the Breusch-Pagan test or White test to statistically confirm heteroscedasticity.
• Remediate the Issue:
	• Target/Feature Transformation: Apply a log transformation or Box-Cox transformation to the skewed target variable (house price), which compresses the scale and stabilizes variance.
	• Weighted Least Squares (WLS): Use WLS regression, which assigns lower weights to observations with higher error variances.
	• Robust Standard Errors: Use heteroscedasticity-consistent (HC / "sandwich") standard errors to fix hypothesis testing and p-values without changing the coefficient estimates themselves.
	• Switch to Non-Linear/Ensemble Models: Transition from basic linear regression to tree-based models like Random Forest or Gradient Boosting (XGBoost/LightGBM), which naturally handle non-linearities and non-constant error distributions better than OLS.

Step 3: Validate and Ensure Robustness

• Cross-Validation: Evaluate the cleaned model using k-fold cross-validation rather than a single train-test split to check generalization.
• Residual Diagnostics: Re-plot the residuals and re-run VIF and Breusch-Pagan/White tests to confirm both issues are mitigated.
'''
```
